```python
#Make Tidy Tables(?)
Precipitation = pd.melt(Precipitation, id_vars='DATE', var_name='City', value_name='Precipitation')
```

```python
#Rename Columns and Values to be easily legible
Precipitation.loc[Precipitation['City'] == 'PRCP_x', 'City'] = 'Seattle'
```

```python
Precipitation.loc[Precipitation['City'] == 'PRCP_y', 'City'] = 'W Lafayette'
```

```python
Precipitation = Precipitation.rename(columns={'DATE': 'Date'})
```

```python
#Impute missing values
Precipitation['Day_of_Year'] = pd.DatetimeIndex(Precipitation['Date']).day_of_year
```

```python
Precipitation
```

```python
SEA_Mean_Precipitation_by_Day = Precipitation.loc[Precipitation['City'] == 'Seattle', ['Precipitation' , 'Day_of_Year']].groupby('Day_of_Year').mean()
```

```python
SEA_Mean_Precipitation_by_Day
```

```python
WLaf_Mean_Precipitation_by_Day = Precipitation.loc[Precipitation['City'] == 'W Lafayette', ['Precipitation' , 'Day_of_Year']].groupby('Day_of_Year').mean()
```

```python
WLaf_Mean_Precipitation_by_Day
```

```python
#Plot Yearly mean rain in each city
plt.figure(figsize=(20,5))
sns.lineplot(data=SEA_Mean_Precipitation_by_Day, x='Day_of_Year', y='Precipitation')
plt.xlabel('Day_of_Year', fontsize=15)
plt.ylabel('Mean Precipitation', fontsize=15)
plt.tick_params(labelsize=17)
plt.show()
```

```python
plt.figure(figsize=(20,5))
sns.lineplot(data=WLaf_Mean_Precipitation_by_Day, x='Day_of_Year', y='Precipitation')
plt.xlabel('Day_of_Year', fontsize=15)
plt.ylabel('Mean Precipitation', fontsize=15)
plt.tick_params(labelsize=17)
plt.show()
```

```python
#Identify where missing values per city
Indices = np.where(Precipitation['Precipitation'].isna() == True)
```

```python
Indices
```

```python
#Replace the missing values with mean values (where applicable)
for index in Indices:
    Precipitation.loc[index, 'Precipitation'] = WLaf_Mean_Precipitation_by_Day.loc[Precipitation.loc[index, 'Day_of_Year']].values[0]
```
